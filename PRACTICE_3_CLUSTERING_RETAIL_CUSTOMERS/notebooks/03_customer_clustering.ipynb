# Practice 3 — Phân cụm khách hàng bán lẻ
Notebook gọi script phân cụm để chạy thống nhất giữa notebook và terminal. Đặt file `Mall_Customers.csv` trong thư mục `data/` và xác minh nguồn/giấy phép trước khi sử dụng.

## Cài đặt
Tại thư mục gốc repository:

```bash
python -m pip install -r PRACTICE_3_CLUSTERING_RETAIL_CUSTOMERS/requirements.txt
```

K-Medoids được cài đặt bằng PAM trong `scripts/cluster.py`; không cần `scikit-learn-extra`.


````python
from pathlib import Path
import runpy
import sys

cwd = Path.cwd().resolve()
project_name = 'PRACTICE_3_CLUSTERING_RETAIL_CUSTOMERS'
candidates = [cwd / project_name, cwd, cwd.parent]
PROJECT = next((p for p in candidates if (p / 'scripts' / 'cluster.py').is_file()), None)
if PROJECT is None:
    raise FileNotFoundError('Không tìm thấy thư mục Lab 3. Mở notebook từ repository root hoặc thư mục Lab 3.')
script = PROJECT / 'scripts' / 'cluster.py'
data = PROJECT / 'data' / 'Mall_Customers.csv'
if not data.is_file():
    raise FileNotFoundError(f'Thiếu dataset: {data}. Xem data/README.md để chuẩn bị dữ liệu.')
sys.argv = [str(script), '--data', str(data)]
runpy.run_path(str(script), run_name='__main__')
````
